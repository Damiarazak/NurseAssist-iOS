# NurseAssist Project - Gantt Chart
## 11-Week Development Timeline for 3-Person Team

This notebook generates a comprehensive Gantt chart for the NurseAssist Flutter project, showing task distribution across three team members over 11 weeks.

## 1. Import Required Libraries

```python
import pandas as pd
import plotly.graph_objects as go
from datetime import datetime, timedelta
import numpy as np

# Suppress warnings
import warnings
warnings.filterwarnings('ignore')
```

## 2. Define Project Tasks and Timeline

```python
# Define the start date (project kickoff - typically a Monday)
project_start = datetime(2026, 1, 26)  # Week 1 Monday

# Define project phases and tasks for NurseAssist
tasks = [
    # Phase 1: Planning & Design (Weeks 1-2)
    {"task": "Project Planning & Requirements", "phase": "Planning", "duration": 2, "order": 1},
    {"task": "UI/UX Design & Wireframes", "phase": "Design", "duration": 2, "order": 2},
    {"task": "Database Schema Design", "phase": "Design", "duration": 1.5, "order": 3},
    
    # Phase 2: Setup & Infrastructure (Weeks 2-3)
    {"task": "Flutter Project Setup", "phase": "Setup", "duration": 1, "order": 4},
    {"task": "Firebase Configuration", "phase": "Setup", "duration": 1, "order": 5},
    {"task": "API & Backend Setup", "phase": "Setup", "duration": 1.5, "order": 6},
    
    # Phase 3: Core Features Development (Weeks 4-8)
    {"task": "Authentication Module", "phase": "Development", "duration": 1.5, "order": 7},
    {"task": "User Profile & Dashboard", "phase": "Development", "duration": 2, "order": 8},
    {"task": "Nurse Scheduling Feature", "phase": "Development", "duration": 2.5, "order": 9},
    {"task": "Patient Records Management", "phase": "Development", "duration": 2.5, "order": 10},
    {"task": "Real-time Notifications", "phase": "Development", "duration": 1.5, "order": 11},
    {"task": "Image & Document Upload", "phase": "Development", "duration": 1.5, "order": 12},
    {"task": "ML Model Integration (TFLite)", "phase": "Development", "duration": 2, "order": 13},
    
    # Phase 4: Testing (Weeks 8-9)
    {"task": "Unit & Widget Testing", "phase": "Testing", "duration": 1.5, "order": 14},
    {"task": "Integration Testing", "phase": "Testing", "duration": 1.5, "order": 15},
    {"task": "UI/UX Testing & Refinement", "phase": "Testing", "duration": 1, "order": 16},
    
    # Phase 5: Optimization & Deployment (Weeks 10-11)
    {"task": "Performance Optimization", "phase": "Optimization", "duration": 1, "order": 17},
    {"task": "Security & Code Review", "phase": "Optimization", "duration": 1, "order": 18},
    {"task": "Bug Fixes & Polish", "phase": "Optimization", "duration": 1, "order": 19},
    {"task": "Deployment & Release", "phase": "Deployment", "duration": 1, "order": 20},
]

print(f"Total tasks defined: {len(tasks)}")
print(f"Project starts: {project_start.strftime('%Y-%m-%d')}")
```

## 3. Create Task Schedule Data

```python
# Calculate start and end dates for each task
schedule_data = []
current_date = project_start

for task in tasks:
    start_date = current_date
    end_date = start_date + timedelta(weeks=task['duration'])
    
    schedule_data.append({
        'Task': task['task'],
        'Phase': task['phase'],
        'Start': start_date,
        'End': end_date,
        'Duration_weeks': task['duration'],
        'Order': task['order']
    })
    
    current_date = end_date

# Create DataFrame
df_schedule = pd.DataFrame(schedule_data)

print("Task Schedule Created:")
print(df_schedule.head(10))
print(f"\nTotal project duration: {(df_schedule['End'].max() - df_schedule['Start'].min()).days / 7:.1f} weeks")
print(f"Project end date: {df_schedule['End'].max().strftime('%Y-%m-%d')}")
```

## 4. Assign Tasks to Team Members

```python
# Assign tasks to team members - rotate assignments to balance workload
# Team members: Developer 1 (Frontend), Developer 2 (Backend), Developer 3 (QA/DevOps)

team_assignment = {
    0: "Dev 1 (Frontend)",      # Project Planning
    1: "Dev 1 (Frontend)",      # UI/UX Design
    2: "Dev 2 (Backend)",       # Database Design
    3: "Dev 1 (Frontend)",      # Flutter Setup
    4: "Dev 2 (Backend)",       # Firebase Config
    5: "Dev 2 (Backend)",       # API Setup
    6: "Dev 1 (Frontend)",      # Authentication
    7: "Dev 1 (Frontend)",      # User Profile & Dashboard
    8: "Dev 1 (Frontend)",      # Nurse Scheduling UI
    9: "Dev 2 (Backend)",       # Patient Records Backend
    10: "Dev 2 (Backend)",      # Real-time Notifications
    11: "Dev 1 (Frontend)",     # Image Upload UI
    12: "Dev 3 (ML/QA)",        # ML Model Integration
    13: "Dev 3 (ML/QA)",        # Unit & Widget Testing
    14: "Dev 3 (ML/QA)",        # Integration Testing
    15: "Dev 3 (ML/QA)",        # UI Testing
    16: "Dev 2 (Backend)",      # Performance Optimization
    17: "Dev 3 (ML/QA)",        # Security Review
    18: "Dev 1 (Frontend)",     # Bug Fixes
    19: "Dev 2 (Backend)",      # Deployment
}

# Add team member assignment to dataframe
df_schedule['Team_Member'] = df_schedule.index.map(team_assignment)

print("Tasks assigned to team members:")
print(df_schedule[['Task', 'Team_Member', 'Duration_weeks']].head(15))
```

## 5. Generate Gantt Chart

```python
# Create Gantt chart using Plotly
fig = go.Figure()

# Define colors for each team member
colors = {
    "Dev 1 (Frontend)": "#1f77b4",      # Blue
    "Dev 2 (Backend)": "#ff7f0e",       # Orange
    "Dev 3 (ML/QA)": "#2ca02c"          # Green
}

# Define colors for each phase
phase_colors = {
    "Planning": "#B3E5FC",
    "Design": "#81D4FA",
    "Setup": "#4FC3F7",
    "Development": "#FFF9C4",
    "Testing": "#FFE082",
    "Optimization": "#FFCC80",
    "Deployment": "#A1887F"
}

# Add bars for each task
for idx, row in df_schedule.iterrows():
    fig.add_trace(go.Bar(
        x=[row['End'] - row['Start']],
        y=[row['Task']],
        base=row['Start'],
        marker=dict(
            color=phase_colors.get(row['Phase'], '#CCCCCC'),
            line=dict(color=colors[row['Team_Member']], width=3)
        ),
        text=f"{row['Team_Member']} ({row['Duration_weeks']:.1f}w)",
        textposition="inside",
        textfont=dict(size=10),
        hovertemplate=f"<b>{row['Task']}</b><br>Team: {row['Team_Member']}<br>Start: %{{base|%Y-%m-%d}}<br>End: %{{customdata|%Y-%m-%d}}<br>Duration: {row['Duration_weeks']:.1f} weeks<extra></extra>",
        customdata=[row['End']],
        orientation='h',
        showlegend=idx < 7,  # Show legend for first few items
        legendgroup=row['Phase'],
        name=row['Phase']
    ))

# Update layout
fig.update_layout(
    title={
        'text': "NurseAssist Project - Gantt Chart (11 Weeks | 3-Person Team)",
        'x': 0.5,
        'xanchor': 'center',
        'font': {'size': 20}
    },
    xaxis=dict(
        type='date',
        title='Timeline',
        gridcolor='#E0E0E0',
        gridwidth=1,
    ),
    yaxis=dict(
        title='Tasks',
        autorange='reversed',
    ),
    barmode='overlay',
    height=1200,
    hovermode='closest',
    plot_bgcolor='#F8F9FA',
    showlegend=True,
    legend=dict(
        x=0.01,
        y=0.99,
        bgcolor='rgba(255, 255, 255, 0.8)',
        bordercolor='#BCCCDC',
        borderwidth=1
    ),
    font=dict(size=11),
    margin=dict(l=250, r=100, t=80, b=80)
)

fig.show()

print("Gantt Chart generated successfully!")
```

## 6. Team Workload Analysis

```python
# Analyze workload distribution
workload_summary = df_schedule.groupby('Team_Member').agg({
    'Duration_weeks': ['sum', 'count'],
    'Task': 'count'
}).round(2)

workload_summary.columns = ['Total_Weeks', 'Task_Count', 'Number_of_Tasks']

print("=" * 60)
print("TEAM WORKLOAD DISTRIBUTION")
print("=" * 60)
print(workload_summary)
print("\n" + "=" * 60)

# Show task breakdown by team member
print("\nTASK BREAKDOWN BY TEAM MEMBER:")
print("=" * 60)
for team in sorted(df_schedule['Team_Member'].unique()):
    team_tasks = df_schedule[df_schedule['Team_Member'] == team]
    print(f"\n{team}:")
    print(f"  Total Duration: {team_tasks['Duration_weeks'].sum():.1f} weeks")
    print(f"  Number of Tasks: {len(team_tasks)}")
    print(f"  Tasks:")
    for _, task in team_tasks.iterrows():
        print(f"    - {task['Task']} ({task['Duration_weeks']:.1f} weeks) [{task['Phase']}]")

# Phase breakdown
print("\n" + "=" * 60)
print("PROJECT PHASES:")
print("=" * 60)
phase_summary = df_schedule.groupby('Phase').agg({
    'Duration_weeks': 'sum',
    'Task': 'count'
}).round(2)
phase_summary.columns = ['Total_Weeks', 'Number_of_Tasks']
print(phase_summary)
```

## 7. Export and Documentation

```python
# Export data to CSV for reference
output_path = "NurseAssist_Project_Schedule.csv"
df_schedule.to_csv(output_path, index=False)
print(f"Schedule exported to: {output_path}")

# Export Gantt chart as HTML
html_path = "NurseAssist_Gantt_Chart.html"
fig.write_html(html_path)
print(f"Gantt chart exported to: {html_path}")

# Create a summary report
print("\n" + "=" * 60)
print("PROJECT SUMMARY")
print("=" * 60)
print(f"Project Name: NurseAssist (Flutter Application)")
print(f"Team Size: 3 Members")
print(f"Project Duration: {(df_schedule['End'].max() - df_schedule['Start'].min()).days / 7:.1f} weeks")
print(f"Start Date: {df_schedule['Start'].min().strftime('%A, %B %d, %Y')}")
print(f"End Date: {df_schedule['End'].max().strftime('%A, %B %d, %Y')}")
print(f"Total Tasks: {len(df_schedule)}")
print(f"\nTeam Composition:")
print(f"  • Dev 1 (Frontend) - Flutter UI/UX Development")
print(f"  • Dev 2 (Backend) - API, Database, and Infrastructure")
print(f"  • Dev 3 (ML/QA) - Testing, ML Integration, and DevOps")
print("\nKey Phases:")
print(f"  1. Planning & Design (Weeks 1-2)")
print(f"  2. Setup & Infrastructure (Weeks 2-3)")
print(f"  3. Core Features Development (Weeks 4-8)")
print(f"  4. Testing (Weeks 8-9)")
print(f"  5. Optimization & Deployment (Weeks 10-11)")
print("=" * 60)
```
